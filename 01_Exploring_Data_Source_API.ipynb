# Creating and Attaching a custom schema

```python
from pyspark.sql.types import *

USER_SCHEMA = StructType(
    [
        StructField("id", IntegerType()),
        StructField("age", IntegerType()),
        StructField("gender", StringType()),
        StructField("designation", StringType()),
        StructField("salaryid", IntegerType())
    ]
)
```

```python
df2 = spark.read.csv(
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/sample_dataset/users_001.csv",
    header=True,
    schema=USER_SCHEMA
)
```

```python
df2.display()
```

# Handling Bad records
1. PERMISSIVE (DEFAULT MODE)
2. DROPMALFORMED
3. FAILFAST

## 1. PERMISSIVE
Automatically the bad record goes to a column named `_corrupt_record`

```python
spark.read.json(
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/access_logs.json"
).display()
```

## 2. DROPMALFORMED
Bad record will simply be rejected in Drop Malformed

```python
spark.read.json(
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/access_logs.json",
    mode = "DROPMALFORMED"
).display()
```

## 3. FAILFAST
If one record is bad, then the entire file is not loaded at all, and throws an error

```python
spark.read.json(
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/access_logs.json",
    mode = "FAILFAST"
).display()
```

# DataFrame Writer API

```python
df3 = spark.read.csv(
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/sample_dataset/users_001.csv",
    header=True,
    schema=USER_SCHEMA
)
```

We write the file now to a Parquet format

```python
df3.write.save(
    format = "parquet",
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet",
    mode = "overwrite"
)

spark.read.parquet(
    "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet"
).display()
```

We again read the saved parquet file

```python
spark.read.parquet(
    "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet"
).display()
```

# Modes of Writing
