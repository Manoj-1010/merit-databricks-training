# Disadvantages of Cloud Storage
1. Does not support ACID Properties
2. Does not support CRUD Operations
3. No versioning support
4. No Time Travel support
5. Have lot of data quality issues

# Lakehouse Architecture
- An Architectural pattern which simply combines the best features of data lake and data warehouse

## Data Lake
- Stores structured, unstructured and semi-structured data

## Data Warehouse
- Stores Structured data
- Supports ACID properties and CRUD operations
