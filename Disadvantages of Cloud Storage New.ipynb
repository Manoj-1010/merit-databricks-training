# Disadvantages of Cloud Storage
1. Does not support ACID Properties
2. Does not support CRUD Operations
3. No versioning support
4. No Time Travel support
5. Have lot of data quality issues

# Lakehouse Architecture
- An Architectural pattern which simply combines the best features of data lake and data warehouse

## Data Lake
- Stores structured, unstructured and semi-structured data

## Data Warehouse
- Stores Structured data
- Supports ACID properties and CRUD operations

# Open Table Format (OTF)
- Brings metadata layer for actual data
- We can use Delta for implementing OTF
- Several OTFs inlude Apache Iceberg, Apache Hudi
- This brings ACID properties, CRUD operations and other advantages

# Prerequisites

## Load `users_001.csv` into DataFrame

```python
df = spark.read.csv(
    path = "/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",
    header = True,
    inferSchema = True
)

df.limit(4).display()
```

## Write to Parquet file

```python
df.write.format("parquet").save(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet",
    mode="overwrite",
)
df_parquet = spark.read.parquet(
    "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet"
)
df_parquet.limit(4).display()
```

## Create a view

```python
df_parquet.createOrReplaceTempView("user_vw")
```

## Drawback - Updates are not allowed

```python
%sql
UPDATE user_vw
SET country = "Bharat"
WHERE country = "India"
```

# Disadvantage - Overwrite

```python
from pyspark.sql.functions import col

df.filter(col("country") == "India").write.format("parquet").save(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet",
    mode="overwrite",
)
```

```python
spark.read.parquet(
    "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet"
).limit(4).display()
```

The entire file is getting replaced without any notification, and this is a serious data quality issue
