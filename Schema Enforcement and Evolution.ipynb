# Prerequisites

## Loading the data into dataframe

```python
df = spark.read.csv(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
)
df.limit(4).display()
```

# Writing the data to the table

```python
df.write.saveAsTable("merit_catalog.quickstart_schema.users_new")
```

```python
spark.read.table("merit_catalog.quickstart_schema.users_new").display()
```

# Schema Enforcement

```python
spark.read.csv(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_006_new_column_education.csv",
    header=True,
    inferSchema=True,
).write.mode("append").saveAsTable("merit_catalog.quickstart_schema.users_new")
```

# Schema Evolution

```python
spark.read.csv(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_006_new_column_education.csv",
    header=True,
    inferSchema=True,
).write.option("mergeSchema",True).mode("append").saveAsTable("merit_catalog.quickstart_schema.users_new")
```

# Checking for null values for append mode

First five columns would have null for education column

```python
spark.read.table("merit_catalog.quickstart_schema.users_new").limit(5).display()
```

```python
from pyspark.sql.functions import col

spark.read.table("merit_catalog.quickstart_schema.users_new").filter(
    col("id").between(2501, 2505)
).display()
```

When we have injected less columns, other column's vales would become null

```python
spark.read.table("merit_catalog.quickstart_schema.users_new").filter(
    col("id").between(5501, 5505)
).display()
```

```python
spark.read.csv(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_014_less_columns_with_different_column_names.csv",
    header=True,
    inferSchema=True,
).select(col("u_id").alias("id")).write.mode("append").saveAsTable(
    "merit_catalog.quickstart_schema.users_new"
)
```
